```python
# Calculate the average target using training data only.
baseline_value = train_df[TARGET].mean()

print(f"Baseline prediction: ₹{baseline_value:,.2f}")
```

```python
def evaluate_predictions(actual, predicted):
    actual = np.asarray(actual, dtype=float)
    predicted = np.asarray(predicted, dtype=float)

    valid = (
        np.isfinite(actual)
        & np.isfinite(predicted)
    )

    actual = actual[valid]
    predicted = predicted[valid]

    return {
        "Rows Evaluated": len(actual),
        "MAE (₹)": mean_absolute_error(actual, predicted),
        "RMSE (₹)": np.sqrt(
            mean_squared_error(actual, predicted)
        )
    }


baseline_results = {}

for name, df in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    predictions = np.full(len(df), baseline_value)

    baseline_results[name] = evaluate_predictions(
        df[TARGET],
        predictions
    )

baseline_results_df = pd.DataFrame(baseline_results).T

display(baseline_results_df.round(2))
```

```python
OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

baseline_results_df.to_csv(
    OUTPUT_DIR / "baseline_results.csv",
    index_label="Dataset"
)

print("Saved:", OUTPUT_DIR / "baseline_results.csv")
```

```python
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

DATA_DIR = Path("data/processed")
TARGET = "target_next_month_expenses"

train_df = pd.read_csv(DATA_DIR / "forecast_ml_train.csv")
val_df = pd.read_csv(DATA_DIR / "forecast_ml_validation.csv")

# Prepare features
excluded = ["user_id", TARGET]

X_train = train_df.drop(columns=excluded, errors="ignore")
y_train = train_df[TARGET]

X_val = val_df.drop(columns=excluded, errors="ignore")
y_val = val_df[TARGET]

# Calculate the upper IQR threshold from training targets only
q1 = y_train.quantile(0.25)
q3 = y_train.quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr

# Baseline: train on all observations
baseline_model = LinearRegression()
baseline_model.fit(X_train, y_train)
baseline_pred = baseline_model.predict(X_val)

# Experiment: exclude high-target observations from training only
keep = y_train <= upper_bound

filtered_model = LinearRegression()
filtered_model.fit(X_train.loc[keep], y_train.loc[keep])
filtered_pred = filtered_model.predict(X_val)

# Compare on the SAME, unchanged validation set
results = pd.DataFrame([
    {
        "Model": "All training observations",
        "Training rows": len(y_train),
        "Validation MAE": mean_absolute_error(y_val, baseline_pred),
        "Validation RMSE": root_mean_squared_error(y_val, baseline_pred),
    },
    {
        "Model": "Exclude high-target training observations",
        "Training rows": int(keep.sum()),
        "Validation MAE": mean_absolute_error(y_val, filtered_pred),
        "Validation RMSE": root_mean_squared_error(y_val, filtered_pred),
    },
])

print(f"Training upper IQR bound: ₹{upper_bound:,.2f}")
print(f"Training observations flagged: {(~keep).sum()}")
print(results.to_string(index=False))
```

```python
import pandas as pd

from pathlib import Path

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
from sklearn.linear_model import LinearRegression

DATA_DIR = Path("data/processed")
TARGET = "target_next_month_expenses"

train_df = pd.read_csv(DATA_DIR / "forecast_ml_train.csv")
val_df = pd.read_csv(DATA_DIR / "forecast_ml_validation.csv")

# Prepare features
excluded = ["user_id", TARGET]

X_train = train_df.drop(columns=excluded, errors="ignore")
y_train = train_df[TARGET]

X_val = val_df.drop(columns=excluded, errors="ignore")
y_val = val_df[TARGET]

# Baseline: Linear Regression
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_pred = linear_model.predict(X_val)

# Nonlinear model: Extra Trees
extra_trees = ExtraTreesRegressor(
    n_estimators=300,
    max_depth=None,
    min_samples_split=5,
    min_samples_leaf=2,
    max_features=1.0,
    random_state=42,
    n_jobs=-1
)

extra_trees.fit(X_train, y_train)

extra_pred = extra_trees.predict(X_val)

# Compare models
results = pd.DataFrame([
    {
        "Model": "Linear Regression",
        "Validation MAE": mean_absolute_error(y_val, linear_pred),
        "Validation RMSE": root_mean_squared_error(y_val, linear_pred)
    },
    {
        "Model": "Extra Trees",
        "Validation MAE": mean_absolute_error(y_val, extra_pred),
        "Validation RMSE": root_mean_squared_error(y_val, extra_pred)
    }
])

print(results.to_string(index=False))
```
